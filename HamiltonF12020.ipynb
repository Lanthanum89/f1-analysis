# Lewis Hamilton's 2020 Season Race Wins Analysis

In 2020, Lewis Hamilton had a dominant season, winning his 7th World Championship. Let's analyse his race wins throughout that season.

```python
# Import additional libraries for data analysis
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
import fastf1
import fastf1.plotting

# Set up plotting
fastf1.plotting.setup_mpl(color_scheme='fastf1')

# Get all 2020 race sessions
year = 2020
races_2020 = []

# 2020 F1 calendar (race weekends)
race_schedule = [
    'Austria', 'Styria', 'Hungary', 'Britain', 'Spain', 
    'Belgium', 'Italy', 'Tuscany', 'Russia', 'Eifel',
    'Portugal', 'Imola', 'Turkey', 'Bahrain', 'Abu Dhabi'
]

print("Loading 2020 F1 season data...")
print("This may take a few minutes as we fetch data for all races...")
```

```python
# Function to get race results for a specific race
def get_race_winner(year, race_name):
    try:
        session = fastf1.get_session(year, race_name, 'R')
        session.load()
        results = session.results
        winner = results.iloc[0]  # First position is the winner
        return {
            'race': race_name,
            'winner_abbreviation': winner['Abbreviation'],
            'winner_name': f"{winner['FirstName']} {winner['LastName']}",
            'team': winner['TeamName'],
            'time': winner['Time']
        }
    except Exception as e:
        print(f"Error loading {race_name}: {e}")
        return None

# Collect race winners for 2020 season
hamilton_wins = []
all_race_results = []

print("Analyzing race winners...")
```

```python
# Process each race in the 2020 season
for i, race in enumerate(race_schedule, 1):
    print(f"Processing Race {i}: {race}")
    result = get_race_winner(2020, race)
    if result:
        all_race_results.append(result)
        # Check if Hamilton won (his abbreviation is 'HAM')
        if result['winner_abbreviation'] == 'HAM':
            hamilton_wins.append(result)
            print(f"  ✓ Hamilton WIN at {race}")
        else:
            print(f"  - Won by {result['winner_name']} ({result['team']})")

print(f"\nLewis Hamilton won {len(hamilton_wins)} out of {len(all_race_results)} races in 2020")
```

```python
# Create a visualization of Hamilton's wins vs other winners
hamilton_win_races = [result['race'] for result in hamilton_wins]
other_winners = [result for result in all_race_results if result['winner_abbreviation'] != 'HAM']

# Create a summary DataFrame
results_df = pd.DataFrame(all_race_results)
hamilton_wins_df = pd.DataFrame(hamilton_wins)

print("Hamilton's 2020 Race Wins:")
print("=" * 40)
for i, win in enumerate(hamilton_wins, 1):
    print(f"{i:2d}. {win['race']}")

print(f"\nWin Rate: {len(hamilton_wins)}/{len(all_race_results)} = {len(hamilton_wins)/len(all_race_results)*100:.1f}%")
```

```python
# Create visualizations
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Bar chart of Hamilton wins vs others
winners_count = results_df['winner_name'].value_counts()
colors = ['#00D2BE' if 'Hamilton' in name else '#FF1E00' for name in winners_count.index]

ax1.bar(range(len(winners_count)), winners_count.values, color=colors)
ax1.set_xticks(range(len(winners_count)))
ax1.set_xticklabels(winners_count.index, rotation=45, ha='right')
ax1.set_ylabel('Number of Wins')
ax1.set_title('2020 F1 Season - Race Winners')
ax1.grid(True, alpha=0.3)

# Timeline of Hamilton's wins
race_numbers = list(range(1, len(all_race_results) + 1))
hamilton_win_flags = [1 if result['winner_abbreviation'] == 'HAM' else 0 for result in all_race_results]

ax2.bar(race_numbers, hamilton_win_flags, color='#00D2BE', alpha=0.7)
ax2.set_xlabel('Race Number')
ax2.set_ylabel('Hamilton Win (1=Yes, 0=No)')
ax2.set_title('Hamilton Wins Throughout 2020 Season')
ax2.set_xticks(race_numbers)
ax2.set_xticklabels([result['race'][:3] for result in all_race_results], rotation=45)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Statistical Deep Dive: Hamilton's 2020 Championship Campaign

Lewis Hamilton's 2020 season was historically significant as it equaled Michael Schumacher's record of 7 World Championships. Let's analyze the statistical dominance of this championship-winning campaign in detail.

### Key Statistics:
- **11 race wins** out of 15 races (73.3% win rate)
- This was the highest single-season win rate of Hamilton's career
- Mercedes as a constructor won **13 out of 15 races** (86.7% win rate)
- Hamilton clinched the championship with **3 races to spare**

The 2020 season was shortened due to the COVID-19 pandemic, making Hamilton's dominance even more impressive given the compressed schedule and unique challenges.

```python
# Advanced statistical analysis and visualizations
import seaborn as sns
from datetime import datetime

# Create a comprehensive race-by-race analysis
race_analysis = []
for i, result in enumerate(all_race_results, 1):
    race_analysis.append({
        'Race_Number': i,
        'Race_Name': result['race'],
        'Winner': result['winner_name'],
        'Hamilton_Won': 1 if result['winner_abbreviation'] == 'HAM' else 0,
        'Mercedes_Won': 1 if 'Mercedes' in result['team'] else 0,
        'Championship_Points': 25 if result['winner_abbreviation'] == 'HAM' else 0  # Simplified points calculation
    })

race_df = pd.DataFrame(race_analysis)

# Calculate cumulative statistics
race_df['Hamilton_Cumulative_Wins'] = race_df['Hamilton_Won'].cumsum()
race_df['Hamilton_Win_Percentage'] = (race_df['Hamilton_Cumulative_Wins'] / race_df['Race_Number']) * 100
race_df['Mercedes_Cumulative_Wins'] = race_df['Mercedes_Won'].cumsum()
race_df['Mercedes_Win_Percentage'] = (race_df['Mercedes_Cumulative_Wins'] / race_df['Race_Number']) * 100

print("Race-by-Race Championship Progression:")
print("="*50)
print(race_df[['Race_Number', 'Race_Name', 'Winner', 'Hamilton_Cumulative_Wins', 'Hamilton_Win_Percentage']].to_string(index=False))
```

```python
# Create advanced visualizations
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(20, 12))

# 1. Cumulative wins progression
ax1.plot(race_df['Race_Number'], race_df['Hamilton_Cumulative_Wins'], 
         marker='o', linewidth=3, markersize=8, color='#00D2BE', label='Hamilton')
ax1.plot(race_df['Race_Number'], race_df['Mercedes_Cumulative_Wins'], 
         marker='s', linewidth=2, markersize=6, color='#C0C0C0', label='Mercedes Total')
ax1.set_xlabel('Race Number')
ax1.set_ylabel('Cumulative Wins')
ax1.set_title('Cumulative Wins Progression Throughout 2020 Season', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend()
ax1.set_xticks(range(1, 16))

# 2. Win percentage evolution
ax2.plot(race_df['Race_Number'], race_df['Hamilton_Win_Percentage'], 
         marker='o', linewidth=3, markersize=8, color='#00D2BE')
ax2.set_xlabel('Race Number')
ax2.set_ylabel('Win Percentage (%)')
ax2.set_title('Hamilton Win Percentage Evolution', fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.set_ylim(0, 100)
ax2.set_xticks(range(1, 16))

# 3. Race results heatmap
win_matrix = race_df['Hamilton_Won'].values.reshape(3, 5)  # 3 rows, 5 columns
race_names_matrix = race_df['Race_Name'].values.reshape(3, 5)
im = ax3.imshow(win_matrix, cmap='RdYlGn', aspect='auto', vmin=0, vmax=1)
ax3.set_title('Hamilton Wins Heatmap (Green=Win, Red=Loss)', fontsize=14, fontweight='bold')
ax3.set_xticks(range(5))
ax3.set_yticks(range(3))
for i in range(3):
    for j in range(5):
        idx = i * 5 + j
        if idx < len(race_names_matrix.flatten()):
            ax3.text(j, i, race_names_matrix[i, j][:3], ha='center', va='center', 
                    fontsize=10, fontweight='bold', 
                    color='white' if win_matrix[i, j] == 0 else 'black')

# 4. Comparison with competitors
competitor_wins = results_df['winner_name'].value_counts()
colors_pie = ['#00D2BE' if 'Hamilton' in name else '#FF1E00' if 'Verstappen' in name 
              else '#C0C0C0' if 'Bottas' in name else '#FFA500' for name in competitor_wins.index]

wedges, texts, autotexts = ax4.pie(competitor_wins.values, labels=competitor_wins.index, 
                                   autopct='%1.1f%%', colors=colors_pie, startangle=90)
ax4.set_title('2020 Season Win Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```
